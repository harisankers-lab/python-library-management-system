```python
books_data = {
    "B101": {"title": "Python Basics", "author": "Anu", "copies": 2},
    "B102": {"title": "Data Science 101", "author": "Binu", "copies": 1},
    "B103": {"title": "AWS Guide", "author": "Cathy", "copies": 3},
    "B104": {"title": "Statistics Made Easy", "author": "Dev", "copies": 1},
}

members_data = [
    {"id": "M1", "name": "Anu", "type": "student"},
    {"id": "M2", "name": "Binu", "type": "teacher"},
    {"id": "M3", "name": "Cathy", "type": "student"},
]
```

```python
class Book:
  def __init__(self, book_id, title, author, copies):
    self.book_id = book_id
    self.title = title
    self.author = author
    self.__copies = copies

  def available_copies(self):
    return self.__copies

  def issue(self):
    if self.__copies > 0:
      self.__copies -= 1
      return True
    return False

  def give_back(self):
    self.__copies += 1

  def __str__(self):
    return f"{self.book_id} | {self.title} by {self.author} | Copies: {self.__copies}"
```

```python
b = Book("B101", "Python Basics", "Anu", 2)
print(b)
print(b.issue(), b.issue(), b.issue())
print(b.available_copies())
b.give_back()
print(b)
```

```python
class Member:
  def __init__(self, member_id, name):
    self.member_id = member_id
    self.name = name
    self.borrowed = []

  def max_books(self):
    return 1

  def member_type(self):
    return "Member"

  def can_borrow(self):
    return len(self.borrowed) < self.max_books()

  def __str__(self):
    return f"{self.member_id} | {self.name} | {self.member_type()} | Borrowed: {len(self.borrowed)}/{self.max_books()}"

class Student(Member):
  def max_books(self):
    return 2

  def member_type(self):
    return "Student"

class Teacher(Member):
  def max_books(self):
    return 4

  def member_type(self):
    return "Teacher"
```

```python
s = Student("M1", "Anu")
t = Teacher("M2", "Binu")
print(s)
print(t)
print(s.can_borrow())
s.borrowed.extend(["B1", "B2"])
print(s.can_borrow())
print(Member("M9", "Guest"))
```

```python
class Library:
  def __init__(self):
    self.books = {}
    self.members = {}

  def load_books(self, data):
    for book_id, info in data.items():
      self.books[book_id] = Book(book_id, info["title"], info["author"], info["copies"])

  def load_members(self, data):
    for item in data:
      if item["type"] == "teacher" :
        self.members[item["id"]] = Teacher(item["id"], item["name"])
      else:
        self.members[item["id"]] = Student(item["id"], item["name"])
  def issue_book(self, member_id, book_id):
    if member_id not in self.members:
      return "Member not found"
    if book_id not in self.books:
      return "Book not found"

    member = self.members[member_id]
    book = self.books[book_id]

    if not member.can_borrow():
      return f"{member.name} has reached the limit"

    if not book.issue():
      return f"{book.title} is not available"

    member.borrowed.append(book_id)
    return f"{member.name} issued {book.title}"
  def return_book(self, member_id, book_id):
    if member_id not in self.members or book_id not in self.books:
      return "Cannot return"

    member = self.members[member_id]
    book = self.books[book_id]

    if book_id not in member.borrowed:
      return "Cannot return"

    member.borrowed.remove(book_id)
    book.give_back()
    return f"{member.name} returned {book.title}"
  def show_books(self):
    for book in self.books.values():
     print(book)

  def show_members(self):
    for member in self.members.values():
      print(member)
```

```python
lib = Library()
lib.load_books(books_data)
lib.load_members(members_data)
print(len(lib.books), len(lib.members))
print(type(lib.members["M2"]).__name__)
print(lib.issue_book("M1", "B101"))
```
